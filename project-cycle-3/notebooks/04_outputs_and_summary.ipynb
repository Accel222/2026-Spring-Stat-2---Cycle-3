# 04 Outputs and Final Summary

This notebook collects the final outputs from Cycle 3 and prepares text for the README, report, and one-slide infographic.

It does not perform new statistical tests.  
It checks files, reads the final result tables, and creates a concise written summary.

## 1. Import libraries and set paths

````python
import pandas as pd
from pathlib import Path
from IPython.display import display, Markdown

project_dir = Path("..")

tables_dir = project_dir / "outputs" / "tables"
figures_dir = project_dir / "outputs" / "figures"
summary_dir = project_dir / "outputs" / "summary"
report_dir = project_dir / "report"
references_dir = project_dir / "references"

summary_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)
references_dir.mkdir(parents=True, exist_ok=True)

print("Project directory:", project_dir.resolve())
````

## 2. Check required output files

This section checks whether the expected tables, figures, and summary files were generated by previous notebooks.

````python
required_files = {
    "Main summary table": tables_dir / "summary_table_alcohol_by_gender.csv",
    "Inference table": tables_dir / "inference_table_alcohol_by_gender.csv",
    "Effect size table": tables_dir / "effect_size_table_alcohol_by_gender.csv",
    "Grade-level summary table": tables_dir / "grade_level_alcohol_by_gender_summary.csv",
    "Grade-level difference table": tables_dir / "grade_level_gender_difference.csv",
    "Main bar chart": figures_dir / "bar_chart_alcohol_by_gender.png",
    "Grade-level bar chart": figures_dir / "grade_level_alcohol_by_gender.png",
    "Grade-level difference chart": figures_dir / "grade_level_gender_difference.png",
    "Final summary": summary_dir / "final_summary.md",
    "Extension summary": summary_dir / "extension_grade_level_summary.md",
}

check_rows = []

for label, path in required_files.items():
    check_rows.append({
        "Item": label,
        "Path": str(path),
        "Exists": path.exists()
    })

check_table = pd.DataFrame(check_rows)
display(check_table)

missing = check_table[check_table["Exists"] == False]

if len(missing) == 0:
    print("All expected files were found.")
else:
    print("Some files are missing. Check the previous notebooks:")
    display(missing)
````

## 3. Load main analysis results

````python
summary_table_path = tables_dir / "summary_table_alcohol_by_gender.csv"
inference_table_path = tables_dir / "inference_table_alcohol_by_gender.csv"
effect_size_table_path = tables_dir / "effect_size_table_alcohol_by_gender.csv"

summary_table = pd.read_csv(summary_table_path)
inference_table = pd.read_csv(inference_table_path)
effect_size_table = pd.read_csv(effect_size_table_path)

print("Summary table:")
display(summary_table)

print("\nInference table:")
display(inference_table)

print("\nEffect size table:")
display(effect_size_table)
````

## 4. Extract key results for final writing

This cell extracts the important values from the output tables.

````python
# Display inference table first to check values
display(inference_table)

# Extract values manually by row order
female_n = int(inference_table.loc[0, "Value"])
female_users = int(inference_table.loc[1, "Value"])
female_prop = float(inference_table.loc[2, "Value"])

male_n = int(inference_table.loc[3, "Value"])
male_users = int(inference_table.loc[4, "Value"])
male_prop = float(inference_table.loc[5, "Value"])

difference = float(inference_table.loc[6, "Value"])
difference_pp = float(inference_table.loc[7, "Value"])

ci_low = float(inference_table.loc[8, "Value"])
ci_high = float(inference_table.loc[9, "Value"])

z_stat = float(inference_table.loc[10, "Value"])
p_value = float(inference_table.loc[11, "Value"])

decision = "reject H0" if p_value < 0.05 else "fail to reject H0"

key_results = {
    "Female n": female_n,
    "Female alcohol users": female_users,
    "Female proportion": female_prop,
    "Male n": male_n,
    "Male alcohol users": male_users,
    "Male proportion": male_prop,
    "Difference": difference,
    "Difference percentage points": difference_pp,
    "CI lower": ci_low,
    "CI upper": ci_high,
    "z statistic": z_stat,
    "p-value": p_value,
    "Decision": decision
}

display(pd.DataFrame(key_results.items(), columns=["Result", "Value"]))
````

## 5. Load extension results

This section loads the grade-level extension tables.

````python
grade_summary_path = tables_dir / "grade_level_alcohol_by_gender_summary.csv"
grade_difference_path = tables_dir / "grade_level_gender_difference.csv"

grade_summary = pd.read_csv(grade_summary_path)
grade_difference = pd.read_csv(grade_difference_path)

print("Grade-level alcohol use by gender:")
display(grade_summary)

print("\nGrade-level gender differences:")
display(grade_difference)
````

## 6. Create a final interpretation text

This text can be used in the README, report, or presentation notes.

````python
final_interpretation = f'''
# Final Interpretation

## Research Question

Is the proportion of current alcohol use different between male and female students?

## Variables

- Group variable: `WhatIsYourSex`
- Response variable: `CurrentAlcoholUse`
- Recoding: `0 = no current alcohol use`, `1 = current alcohol use`

## Method

Because the response variable is binary, a two-proportion z-test was used to compare the current alcohol use proportions between male and female students.

## Main Results

- Female students: n = {female_n}, current alcohol users = {female_users}, proportion = {female_prop * 100:.2f}%
- Male students: n = {male_n}, current alcohol users = {male_users}, proportion = {male_prop * 100:.2f}%
- Difference in proportions, Male - Female = {difference:.4f}, or {difference_pp:.2f} percentage points
- 95% confidence interval for the difference = ({ci_low:.4f}, {ci_high:.4f})
- z statistic = {z_stat:.4f}
- p-value = {p_value:.4f}

## Statistical Conclusion

At alpha = 0.05, we {decision}.

There is not enough statistical evidence to conclude that the proportion of current alcohol use is different between male and female students if the p-value is greater than 0.05.

## Practical Interpretation

Although male students had a slightly higher sample proportion of current alcohol use, the observed difference was small. The estimated difference was only {difference_pp:.2f} percentage points, so the practical gender difference appears limited in this dataset.

## Extension Analysis

As an extension, we compared current alcohol use proportions by sex within each grade level. This descriptive analysis does not replace the main two-proportion z-test. Instead, it provides more context by checking whether the overall male-female pattern is consistent across grades.

The grade-level comparison showed that the gender pattern was not exactly the same in every grade. This means the overall result should be interpreted as an overall comparison, not as proof that the gender pattern is identical across all grade levels.

## Caution

The data are observational survey data. Therefore, the result should be interpreted as association, not causation.
'''

final_path = summary_dir / "final_interpretation.md"
final_path.write_text(final_interpretation, encoding="utf-8")

display(Markdown(final_interpretation))

print("Saved final interpretation to:")
print(final_path)
````
