# Descriptive Statistics and Visualizations

This notebook compares current alcohol use between male and female students using descriptive statistics and visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

project_dir = Path("..")

processed_data_path = project_dir / "data" / "processed" / "cycle3_cleaned_alcohol_gender.csv"
figures_dir = project_dir / "outputs" / "figures"
tables_dir = project_dir / "outputs" / "tables"

figures_dir.mkdir(parents=True, exist_ok=True)
tables_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(processed_data_path)

df.head()
```

```python
summary_table = df.groupby("sex_group").agg(
    sample_size=("current_alcohol_use", "count"),
    alcohol_users=("current_alcohol_use", "sum"),
    alcohol_use_proportion=("current_alcohol_use", "mean")
).reset_index()

summary_table["alcohol_use_percent"] = summary_table["alcohol_use_proportion"] * 100

summary_table
```

```python
summary_table.to_csv(tables_dir / "summary_table_alcohol_by_gender.csv", index=False)

print("Summary table saved.")
```

```python
plt.figure(figsize=(6, 4))

plt.bar(
    summary_table["sex_group"],
    summary_table["alcohol_use_percent"]
)

plt.xlabel("Sex Group")
plt.ylabel("Current Alcohol Use (%)")
plt.title("Current Alcohol Use by Sex Group")
plt.ylim(0, 100)

plt.tight_layout()
plt.savefig(figures_dir / "bar_chart_alcohol_by_gender.png", dpi=300)
plt.show()
```

## Descriptive Interpretation

The descriptive results show that 44.58% of female students currently used alcohol, while 45.77% of male students currently used alcohol.

The observed difference was small. Male students had about 1.19 percentage points higher current alcohol use than female students in this sample.

The bar chart visually compares the two proportions. However, a formal two-proportion z-test is needed to determine whether this observed difference is statistically significant.
